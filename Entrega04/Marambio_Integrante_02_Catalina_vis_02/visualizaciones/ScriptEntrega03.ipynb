```python
df4 = pd.read_csv("basetasasanuales.csv", decimal=',')
# Add a column for bar size to control width
df4['bar_size'] = [600, 300, 180] # Assign decreasing sizes for each bar, now even wider
df4.head()
```

```python
alt.Chart(df4).mark_bar().encode(
    x='Períodos',
    y='Tasa de presidentas electas por año',
    color=alt.Color('Períodos', scale=alt.Scale(scheme='blues', reverse=True)), # Use a sequential color scheme for similar shades, reversed for darkest to lightest
    size=alt.Size('bar_size', type='quantitative', scale=alt.Scale(range=[20, 100]), legend=None) # Adjust pixel range for bar width and remove legend
).properties(
    width=400, # Set the overall chart width
    title='El aumento exponencial de mujeres electas para la presidencia de partidos políticos en Chile en los últimos 37 años'
)
```

```python
# Save the last chart (Chart 4) as an HTML file
alt.Chart(df4).mark_bar().encode(
    x='Períodos',
    y='Tasa de presidentas electas por año',
    color=alt.Color('Períodos', scale=alt.Scale(scheme='blues', reverse=True)),
    size=alt.Size('bar_size', type='quantitative', scale=alt.Scale(range=[20, 100]), legend=None)
).properties(
    width=400,
    title='El aumento exponencial de mujeres electas para la presidencia de partidos políticos en Chile en los últimos 37 años'
).save('chart4_tasa_presidentas_electas.html')
```
