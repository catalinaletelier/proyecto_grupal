```python
import pandas as pd
import altair as alt
```

```python
from google.colab import files
upload = files.upload()
```

```python
df = pd.read_csv("hojatotalcongreso.csv")
```

```python
df
```

```python
df.columns = df.columns.str.strip()
```

```python
alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Total Congreso:Q"
)
```

```python
alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Total Congreso:Q",
    color="Año:N"
)
```

```python
alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Total Congreso:Q",
    color="Año:N"
).properties(
    title="Mujeres en el Congreso en cada elección"
)
```

```python
alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Total Congreso:Q",
    tooltip=["Total Congreso", "Año"],
    color="Año:N"
).properties(
    title="Mujeres en el Congreso en cada elección"
)
```

```python
chart = alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Total Congreso:Q",
    tooltip=["Total Congreso", "Año"],
    color="Año:N"
).properties(
    title="Mujeres en el Congreso en cada elección"
)

chart.save('mujeres_congreso.html')
```
