```python
df = pd.read_csv("Base datos gabinetes ministeriales  - Hoja 1-2.csv")
```

```python
df.head()
```

```python
alt.Chart(df).mark_bar().encode(
    x="Año:N",
    y="Partido:Q")
```

```python
chart = alt.Chart(df).mark_bar().encode(
    x=alt.X("Año:N", title="Año"),
    y=alt.Y("count():Q", title="Número de Registros"),
    color=alt.Color("Partido:N", title="Partido")
).properties(
    title="Distribución de Partidos por Año"
)
chart
```
